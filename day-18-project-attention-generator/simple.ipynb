## The Question

## The Stack (Read This Top to Bottom)

```
"to be or not to be"               ← raw text
       ↓ tokenize (Day 4)
[12, 8, 1, 9, 18, 17, 1, ...]      ← token IDs (small integers)
       ↓ token embedding (Day 11)
[[0.1, 0.5, ...],
 [0.3, -0.2, ...],
  ...]                              ← each token is now a vector
       ↓ + position embedding (Day 17)
[[0.1, 0.5, ...]+pos0,
 [0.3, -0.2, ...]+pos1, ...]       ← each token also knows WHERE it is
       ↓ multi-head causal attention (Days 15+16)
[updated vectors after attention]   ← each token looks at past tokens
       ↓ output linear layer
[logits for next token at each position]
       ↓ softmax + sample
[next predicted token]
```

5 layers of transformation. That's a complete language model.

## Why Combine Everything Today?

You've built each piece in isolation. But ML isn't real until you can:

1. **Run the whole pipeline end-to-end** — one mistake breaks everything
2. **Track val loss properly** — does it actually work on data it hasn't seen?
3. **Save and reload** — useless if you can't reuse the model
4. **Generate from a prompt** — the user-facing API

Today is the practice. You assemble all the parts into one clean, polished project.

## The Generation Loop

Once trained, you generate text like this:

```
1. Start with a prompt: "to be"
2. Pass it through the model → get next-token probabilities
3. Sample a token (with temperature / top-K)
4. Append to the prompt
5. Repeat
```

This same loop runs in ChatGPT every time it streams a response. Just at much bigger scale.

## Sampling Tricks Worth Knowing

```
ARGMAX (greedy):    Always take the most likely token.
                    Boring, repetitive.

TEMPERATURE = T:    Divide logits by T before softmax.
                    T=0.5 → confident, predictable
                    T=1.0 → normal
                    T=1.5 → creative, weird

TOP-K = N:          Only sample from the N most likely tokens.
                    Prevents weird low-probability picks.

TOP-P = p:          Sample from smallest set with total prob ≥ p.
                    Adaptive — wider when unsure, narrower when sure.
```

Real LLM APIs (OpenAI, Anthropic) expose these knobs to you.
