# Day 12 — Simplified: Building a Real ML Project

## The Question

You've built sentiment classifiers in pieces. **What does it look like to put them all together as a real project?**

## What's different about a "project"

A notebook = playground. A project = something you could ship.

```
NOTEBOOK style:                    PROJECT style:
- All globals everywhere           - Clean classes with one job
- Magic numbers buried in code     - One Config object up top
- "Works on my machine"            - Saved artifacts: weights + vocab + config
- One model                         - Compare ≥ 2 models against a baseline
- Pretty graphs                    - Error analysis: which examples fail?
```

## The 6 Steps of Every ML Project

```
1. CONFIGURE   → one place for all hyperparameters
2. TOKENIZE    → text → numbers (with save/load!)
3. SPLIT       → train / val / test (don't cheat!)
4. MODEL       → define what to learn
5. TRAIN       → loop, track val loss, save best
6. INFERENCE   → load saved model, predict on new text
```

Skip any of these and you have a hack, not a project.

## What we'll do today

Build a sentiment classifier (positive / negative movie reviews) that:
- Loads from saved files (model.pth, vocab.json, config.json)
- Has a clean `.predict("any text")` API
- Reports accuracy + shows which examples it got wrong
- Compares 3 different architectures side by side

This is the template you'll reuse for EVERY future text classification task.

## Why compare against a baseline?

In ML, you always need to ask: "Is my fancy model actually doing better than the simplest possible model?"

```
BASELINE:    Bag of Words + 1 Linear layer       (Day 10)
TARGET:      Embedding + Mean pool + Linear       (Day 11)
STRETCH:     Embedding + 2 hidden layers          (deeper)
```

If your "fancy" model gets 90% and your dumb baseline gets 89%, the fancy one isn't really helping.

## Why error analysis matters

Looking at WHERE your model fails tells you what's broken:

```
Looking at wrong predictions:
- "This was not amazing." → predicted POSITIVE (wrong!)
  → Why? Model sees "amazing", ignores "not" because BoW has no order.
  → Fix: use sequence models (Day 13+)

- "Cinematographic genius..." → predicted NEGATIVE (wrong!)
  → Why? "Cinematographic" wasn't in training vocab → becomes <unk>
  → Fix: bigger vocab or subword tokenization (Day 21)
```

Look at the failures, not just the accuracy number. That's where you learn what to improve.
